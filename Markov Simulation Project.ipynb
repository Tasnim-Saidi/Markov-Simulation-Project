```python
# lire le nom du fichier uploadé
nom_fichier = list(uploaded.keys())[0]

with open(nom_fichier, "r", encoding="utf-8") as f:
    mots = [ligne.strip().lower() for ligne in f if ligne.strip()]

print("Nombre de mots dans le dictionnaire :", len(mots))
print("Exemple :", mots[:10])
```

```python
# ajouter symbole fin de mot
mots = [mot + "#" for mot in mots]

# alphabet utilisé
alphabet = list("abcdefghijklmnopqrstuvwxyz#")

print(alphabet)
```

```python
compteur_initial = {lettre: 0 for lettre in alphabet[:-1]}

for mot in mots:
    premiere = mot[0]
    if premiere in compteur_initial:
        compteur_initial[premiere] += 1

total_initial = sum(compteur_initial.values())

pi0 = np.array([compteur_initial[l]/total_initial for l in alphabet[:-1]])

print("Distribution initiale π(0) :")
for i, l in enumerate(alphabet[:-1]):
    print(l, ":", round(pi0[i],4))
```

```python
n = len(alphabet)
P = np.zeros((n, n))

for mot in mots:
    for i in range(len(mot)-1):
        a = mot[i]
        b = mot[i+1]

        if a in alphabet and b in alphabet:
            ia = alphabet.index(a)
            ib = alphabet.index(b)
            P[ia, ib] += 1

# normalisation des probabilités
for i in range(n):
    s = np.sum(P[i])
    if s != 0:
        P[i] = P[i] / s

print("Matrice de transition créée avec succès.")
print(P)
```

```python
def generer_mot():
    # première lettre selon distribution initiale
    lettre = np.random.choice(alphabet[:-1], p=pi0)
    mot = lettre

    while lettre != "#":
        i = alphabet.index(lettre)
        suivante = np.random.choice(alphabet, p=P[i])

        if suivante != "#":
            mot += suivante

        lettre = suivante

    return mot
```

```python
print("===== MOTS GÉNÉRÉS =====\n")

for _ in range(20):
    print(generer_mot())
```
